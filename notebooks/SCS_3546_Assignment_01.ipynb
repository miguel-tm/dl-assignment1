```python
def set_global_seed(seed: int = 42) -> None:
    """Set random seeds for reproducible results across Python, NumPy, and TensorFlow."""
    os.environ["PYTHONHASHSEED"] = str(seed)
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)

set_global_seed(42)
```
